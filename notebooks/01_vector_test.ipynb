# Legal AI - Vectors

Bu notebook, sistemin kullandığı embedding modeli olan `intfloat/e5-large` üzerinden denemeler yapmak için eklendi.

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer('intfloat/e5-large')

documents = [
    "passage: Mahkeme, sanığın beraatine karar vermiştir.",
    "passage: Kiracı kirayı geciktirirse tahliye başlatılabilir."
]
query = "query: Kiracı evden nasıl çıkarılır?"

doc_embeddings = model.encode(documents, normalize_embeddings=True)
query_embedding = model.encode([query], normalize_embeddings=True)

scores = (query_embedding @ doc_embeddings.T)
print("Scores:", scores)
```
